```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('zomato.csv',encoding='latin-1')
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.describe()
```

```python
## MISSING VALUES IN DATASET

df.isnull().sum()
```

```python
[features for features in df.columns if df[features].isnull().sum()>0]
```

```python
df_country=pd.read_excel('Country-Code.xlsx')
df_country.head()
```

```python
df_final=pd.merge(df,df_country,on='Country Code',how='left')
```

```python
df_final.columns
```

```python
country_names = df_final['Country'].value_counts().index
```

```python
country_val=df_final['Country'].value_counts().values
```

```python
##PIE CHART- top 5 countries
plt.pie(country_val[:3],labels=country_names[:3],autopct='%1.2f%%')
```

OBSERVATION
Zomato maximum transactions are from India after that USA and then United Kingdom

```python
df_final.columns
```

```python
ratings= df_final.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size().reset_index().rename(columns={0:'Rating Count'})
```

```python
ratings
```

Conculsions
1. When the rating is between 4.5-4.9 ---> Excellent
2. When the rating is between 4.0-4.4 ---> Very Good
3. When the rating is between 3.5-3.9 ---> Good
4. When the rating is between 2.5-3.4 ---> Average
5. When the rating is between 1.8-2.4 ---> Poor



```python
ratings.columns
```

```python
import matplotlib
matplotlib.rcParams['figure.figsize']=(15,10)
sns.barplot(x="Aggregate rating",y="Rating Count",hue="Rating color",data=ratings,palette=['blue','red','orange','yellow','green','green'])
```

Obervations
1. Not Rated count is very high
2. Maximum number of rating are between 2.9 to 3.7

```python
#Count plot
sns.countplot(x='Rating color',data=ratings,palette=['blue','red','orange','yellow','green','green'])
#
```
